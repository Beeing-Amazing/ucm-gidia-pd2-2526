# Path configuration

```python
projectRoot = Path.cwd().parents[1]
projectRoot
```

```python
dataPath = Path(projectRoot, 
                "data/tlc_trip_record/yellow") # Replace with the path of your data
dataPath
```

# Load Data

```python
baseName = "yellow_tripdata_"
```

```python
data = pd.DataFrame()
```

```python
years = range(2024,2026,1)
for year in years:
    print(year, end= ", ")
```

```python
months = [ f"{m:02d}" for m in range(1,13,1)]
months
```

```python
# WARNING, the data might be too big to fit into memory
# a year of yellow tlc data is around 5.7 GB 
for year in years:
    for month in months:
        filePath = Path(dataPath, baseName + f"{year}-{month}.parquet") 
        try:
            d = pd.read_parquet(filePath)
        except FileNotFoundError:
            print(f"File {filePath} doesn't exist")
            continue
        data = pd.concat([data,d])
```
